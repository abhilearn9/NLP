```python
# ============================================================
# CELL 1: IMPORT LIBRARIES AND USER INPUT
# ============================================================

import re
from collections import Counter

# Ask the user to enter a paragraph
# eg:paragraph = I love NLP. NLP is interesting. I love learning it.
paragraph = input("Enter your paragraph: ")

print("\nYour paragraph:")
print(paragraph)
```

```python
# ============================================================
# CELL 2: NUMBER OF WORDS
# ============================================================

# Split the paragraph into individual words
words = paragraph.split()
print(words)
# Count the number of words
num_words = len(words)

print("Number of words:", num_words)
```

```python
# ============================================================
# CELL 3: NUMBER OF SENTENCES
# ============================================================

# Split the paragraph at ., !, or ?
sentences = re.split(r'[.!?]', paragraph)
print(sentences)

# Remove empty sentences and extra spaces
sentences = [s.strip() for s in sentences if s.strip()]
print(sentences)

# Count sentences
num_sentences = len(sentences)

print("Number of sentences:", num_sentences)

# Display each sentence
for i, sentence in enumerate(sentences, 1):
    print("Sentence", i, ":", sentence)
```

```python
# ============================================================
# CELL 4: WORD REPETITION COUNT
# ============================================================

# Convert all words to lowercase
# Remove punctuation
words_lower = [
    word.lower().strip(".,!?")
    for word in words
]

# Count each word
word_freq = Counter(words_lower)

print("Word frequencies:")

for word, count in word_freq.items():
    print(word, "->", count)
```

```python
# ============================================================
# CELL 5: SENTENCE REPETITION
# ============================================================

# Convert sentences to lowercase
sentences_lower = [
    sentence.lower()
    for sentence in sentences
]

# Count each sentence
sentence_freq = Counter(sentences_lower)

print("Repeated sentences:")

found = False

for sentence, count in sentence_freq.items():

    if count > 1:

        print(
            sentence,
            "-> appears",
            count,
            "times"
        )

        found = True

if found == False:
    print("No repeated sentences found.")
```

```python
# ============================================================
# CELL 6: TENSE DETECTION
# ============================================================

def detect_tense(sentence):

    # Convert sentence to lowercase
    sentence_lower = sentence.lower()

    # Split sentence into words
    words = sentence_lower.split()

    # Remove punctuation
    words = [
        word.strip(".,!?")
        for word in words
    ]

    # Future tense
    if (
        "will" in words
        or "shall" in words
        or "going to" in sentence_lower
    ):
        return "Future"

    # Past tense words
    past_words = [
        "was",
        "were",
        "did",
        "went",
        "had",
        "played",
        "saw",
        "loved",
        "learned"
    ]

    if any(word in words for word in past_words):
        return "Past"

    # Present tense words
    present_words = [
        "is",
        "am",
        "are",
        "love",
        "learn",
        "like",
        "play",
        "go",
        "eat"
    ]

    if any(word in words for word in present_words):
        return "Present"

    return "Unknown"


# Check every sentence
for i, sentence in enumerate(sentences, 1):

    print(
        "Sentence",
        i,
        "->",
        detect_tense(sentence)
    )
```

```python
# ============================================================
# CELL 7: SENTIMENT ANALYSIS
# ============================================================

# Positive words
positive_words = [
    "good",
    "great",
    "happy",
    "love",
    "excellent",
    "awesome",
    "interesting",
    "amazing",
    "wonderful"
]

# Negative words
negative_words = [
    "bad",
    "sad",
    "hate",
    "terrible",
    "poor",
    "angry",
    "horrible",
    "boring"
]


def sentiment(sentence):

    # Convert to lowercase
    words = sentence.lower().split()

    # Remove punctuation
    words = [
        word.strip(".,!?")
        for word in words
    ]

    # Count positive words
    positive_count = sum(
        1 for word in words
        if word in positive_words
    )

    # Count negative words
    negative_count = sum(
        1 for word in words
        if word in negative_words
    )

    # Compare counts
    if positive_count > negative_count:
        return "Positive"

    elif negative_count > positive_count:
        return "Negative"

    else:
        return "Neutral"


# Analyze every sentence
for i, sentence in enumerate(sentences, 1):

    print(
        "Sentence",
        i,
        "->",
        sentiment(sentence)
    )
```

```python
# ============================================================
# CELL 8: TONE / IMPRESSION
# ============================================================

casual = [
    "bro",
    "dude",
    "lol",
    "hey",
    "cool"
]

aggressive = [
    "stupid",
    "idiot",
    "hate",
    "worst"
]

formal = [
    "kindly",
    "regarding",
    "sincerely",
    "request",
    "therefore"
]

emotional = [
    "happy",
    "sad",
    "angry",
    "love",
    "upset",
    "excited"
]


def tone(sentence):

    # Convert sentence to lowercase
    words = sentence.lower().split()

    # Remove punctuation
    words = [
        word.strip(".,!?")
        for word in words
    ]

    # Check aggressive words
    if any(word in aggressive for word in words):
        return "Aggressive"

    # Check formal words
    if any(word in formal for word in words):
        return "Formal"

    # Check emotional words
    if any(word in emotional for word in words):
        return "Emotional"

    # Check casual words
    if any(word in casual for word in words):
        return "Casual"

    return "Neutral"


# Analyze every sentence
for i, sentence in enumerate(sentences, 1):

    print(
        "Sentence",
        i,
        "->",
        tone(sentence)
    )
```

```python
# ============================================================
# CELL 9: EMOTION / FEELING
# ============================================================

happiness = [
    "happy",
    "glad",
    "excited",
    "joy",
    "awesome"
]

sadness = [
    "sad",
    "upset",
    "hurt",
    "crying",
    "broken"
]

anger = [
    "angry",
    "mad",
    "furious",
    "hate"
]

fear = [
    "afraid",
    "scared",
    "worried",
    "terrified"
]

surprise = [
    "wow",
    "unbelievable",
    "amazed",
    "shocked"
]

disgust = [
    "disgusting",
    "gross",
    "horrible"
]


def emotion(sentence):

    # Convert sentence to lowercase
    words = sentence.lower().split()

    # Remove punctuation
    words = [
        word.strip(".,!?")
        for word in words
    ]

    # Check happiness
    if any(word in happiness for word in words):
        return "Happiness"

    # Check sadness
    if any(word in sadness for word in words):
        return "Sadness"

    # Check anger
    if any(word in anger for word in words):
        return "Anger"

    # Check fear
    if any(word in fear for word in words):
        return "Fear"

    # Check surprise
    if any(word in surprise for word in words):
        return "Surprise"

    # Check disgust
    if any(word in disgust for word in words):
        return "Disgust"

    return "Neutral"


# Analyze every sentence
for i, sentence in enumerate(sentences, 1):

    print(
        "Sentence",
        i,
        "->",
        emotion(sentence)
    )
```

```python
#Gutenberg
import nltk

nltk.download("gutenberg")
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")

from nltk.corpus import gutenberg
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag


# Select one particular Gutenberg file
file_id = "austen-emma.txt"


# Get the text
text = gutenberg.raw(file_id)


# Tokenization
tokens = word_tokenize(text)


# Stopwords
stop_words = set(stopwords.words("english"))


# Remove stopwords
filtered_words = [
    word
    for word in tokens
    if word.lower() not in stop_words
]


# POS tagging
tagged_words = pos_tag(filtered_words)


# Assign integer to every unique word
word_numbers = {}

number = 1

for word, tag in tagged_words:

    if word.lower() not in word_numbers:

        word_numbers[word.lower()] = number

        number += 1


# Display result
print("Integer   Word   POS")

for word, tag in tagged_words[:50]:

    print(
        word_numbers[word.lower()],
        word,
        tag
    )
```

```python
#NPS Chat Corpus

#NPS Chat is a little different because it contains chat messages.

#First download it:

import nltk

nltk.download("nps_chat")
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")

#Import:

from nltk.corpus import nps_chat
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag
```

```python
#the available files
print(nps_chat.fileids())

#Select one particular file
file_id = nps_chat.fileids()[0]

print("Selected file:", file_id)
#Get the posts
posts = nps_chat.posts(file_id)

print(posts[:10])
```

```python
#Combine the posts

#We can combine all the words into one list.

words = []

for post in posts:

    for word in post:

        words.append(word)

#Check:

print(words[:50])
```

```python
#Remove stopwords
stop_words = set(stopwords.words("english"))

filtered_words = [
    word
    for word in words
    if word.lower() not in stop_words
]
```

```python
#POS tagging
tagged_words = pos_tag(filtered_words)

print(tagged_words[:30])
```

```python
#Assign integer to every word
word_numbers = {}

number = 1

for word, tag in tagged_words:

    if word.lower() not in word_numbers:

        word_numbers[word.lower()] = number

        number += 1

#Display:

print("Integer   Word   POS")

for word, tag in tagged_words[:50]:

    print(
        word_numbers[word.lower()],
        word,
        tag
    )
```
