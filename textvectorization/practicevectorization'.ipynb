```python
from sklearn.feature_extraction.text import TfidfVectorizer
sentences=['The young student is reading a Python book in the classroom',
'The young student is writing Python code on the computer',
'The teacher is reading the Python book while the students are writing notes',]
#'The students are learning Python in the classroom with their teacher']
```

```python
vectors=TfidfVectorizer()
vectors.fit(sentences)

transform=vectors.transform(sentences)
print(transform)
print(vectors.vocabulary_)
print(transform.shape)
```

```python
#binary
vectors=TfidfVectorizer(use_idf=True)
vectors.fit(sentences)
transform=vectors.transform(sentences)
print(transform)
print(transform.toarray())
sentences=['The young student is reading a Python book in the classroom',
'The young student is writing Python code on the computer',
'The teacher is reading the Python book while the students are writing notes',
'The students are learning Python in the classroom with their teacher']
```
