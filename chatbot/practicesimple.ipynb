```python
import json
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# Load the JSON file
with open("/content/drive/MyDrive/Natural Language Processing/chatbot/data.json",
          "r", encoding="utf-8") as file:

    data = json.load(file)


# Get questions from JSON
questions = []

for item in data:
    questions.append(item["Question"])


# Create TF-IDF
vectorizer = TfidfVectorizer()

question_vectors = vectorizer.fit_transform(questions)


# This function finds the answer
def find_answer(user_question):

    # Convert user's question
    user_vector = vectorizer.transform([user_question])

    # Compare with all questions
    scores = cosine_similarity(
        user_vector,
        question_vectors
    )[0]

    # Find highest score
    best_index = scores.argmax()

    # Get highest score
    best_score = scores[best_index]

    # Check the score
    if best_score >= 0.65:

        answer = data[best_index]["Answer"]

        return answer, best_score


    elif best_score >= 0.30:

        return "I found a similar question: " + questions[best_index], best_score


    else:

        return "Sorry, I could not understand your question.", best_score


# Start chatbot
print("Chatbot started!")
print("Type 'exit' to stop.")


while True:

    user_question = input("\nYou: ")

    if user_question.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    answer, score = find_answer(user_question)

    print("Chatbot:", answer)
    print("Similarity:", round(score, 3))
```
